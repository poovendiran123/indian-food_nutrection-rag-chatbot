```python
datasets = {
    "Food Recipes": food_recipes,
    "Fruits": fruits,
    "Meat": meat,
    "Seafood": seafood,
    "Vegetables": vegetables
}

for name, records in datasets.items():

    print(f"\n{'='*50}")
    print("Dataset:", name)
    print("Total Records:", len(records))

    if len(records) > 0:
        print("Available Columns:", list(records[0].keys()))
```

```python
import json

CHUNK_SIZE = 1000
CHUNK_OVERLAP = 100

assert CHUNK_SIZE > CHUNK_OVERLAP >= 0

chunks = []

for category, records in datasets.items():

    for record_index, record in enumerate(records):

        # Get food or recipe name
        record_name = (
            record.get("name")
            or record.get("clean_name")
            or record.get("food_name")
            or f"{category}_{record_index}"
        )

        # Convert record to JSON text
        text = json.dumps(
            record,
            ensure_ascii=False,
            indent=2
        )

        start = 0
        chunk_index = 0

        while start < len(text):

            end = min(start + CHUNK_SIZE, len(text))

            chunk_text = text[start:end]

            # Add record context to every chunk
            content = (
                f"Category: {category}\n"
                f"Food Name: {record_name}\n\n"
                f"{chunk_text}"
            )

            chunks.append({
                "id": f"{category}_{record_index}_{chunk_index}",
                "category": category,
                "record_id": record_index,
                "record_name": record_name,
                "chunk_id": chunk_index,
                "content": content
            })

            if end >= len(text):
                break

            start = end - CHUNK_OVERLAP
            chunk_index += 1


print("Chunk Size:", CHUNK_SIZE)
print("Chunk Overlap:", CHUNK_OVERLAP)
print("Total Chunks:", len(chunks))

print("\nSample Chunk:")
print(chunks[0]["content"])
```

```python
from collections import Counter
from statistics import mean

# 1. Total chunks
print("Total Chunks:", len(chunks))

# 2. Category-wise chunks
print("\nCategory-wise Chunk Count:")
category_counts = Counter(c["category"] for c in chunks)

for category, count in category_counts.items():
    print(category, ":", count)

# 3. Chunk length statistics
lengths = [len(c["content"]) for c in chunks]

print("\nChunk Length Statistics:")
print("Minimum:", min(lengths))
print("Maximum:", max(lengths))
print("Average:", round(mean(lengths), 2))

# 4. Empty chunks
empty_chunks = sum(
    not c["content"].strip() for c in chunks
)
print("\nEmpty Chunks:", empty_chunks)

# 5. Duplicate content
contents = [c["content"] for c in chunks]
print("Duplicate Chunks:", len(contents) - len(set(contents)))

# 6. Duplicate IDs
chunk_ids = [
    (c["category"], c["record_id"], c["chunk_id"])
    for c in chunks
]

print("Duplicate Chunk IDs:",
      len(chunk_ids) - len(set(chunk_ids)))
```

```python
from collections import defaultdict

content_map = defaultdict(list)

for chunk in chunks:

    content_map[chunk["content"]].append({
        "category": chunk["category"],
        "record_id": chunk["record_id"],
        "record_name": chunk["record_name"],
        "chunk_id": chunk["chunk_id"]
    })

duplicate_groups = {
    content: records
    for content, records in content_map.items()
    if len(records) > 1
}

print("Duplicate Groups:", len(duplicate_groups))

for i, (content, records) in enumerate(
    list(duplicate_groups.items())[:10], 1
):

    print(f"\n{'='*50}")
    print("Duplicate Group:", i)
    print("Occurrences:", len(records))
    print("Records:", records)
    print("Content Preview:", content[:300])
```

```python
exact_duplicates = 0
different_records = 0

for group_no, (content, locations) in enumerate(
    duplicate_groups.items(), 1
):

    source_records = []

    for item in locations:
        record = datasets[item["category"]][item["record_id"]]
        source_records.append(record)

    # Compare original JSON records
    is_exact_duplicate = all(
        record == source_records[0]
        for record in source_records[1:]
    )

    if is_exact_duplicate:
        exact_duplicates += 1
    else:
        different_records += 1

    if group_no <= 10:
        print("\n" + "=" * 50)
        print("Duplicate Group:", group_no)
        print("Food Name:", locations[0]["record_name"])
        print("Occurrences:", len(locations))
        print("Exact Original Record Duplicate:", is_exact_duplicate)

print("\n========== FINAL SUMMARY ==========")
print("Total Duplicate Groups:", len(duplicate_groups))
print("Exact Duplicate Records:", exact_duplicates)
print("Different Record Groups:", different_records)
```

```python
import json

datasets_dedup = {}

for category, records in datasets.items():

    seen = set()
    unique_records = []

    for record in records:

        record_key = json.dumps(
            record,
            sort_keys=True,
            ensure_ascii=False
        )

        if record_key not in seen:
            seen.add(record_key)
            unique_records.append(record)

    datasets_dedup[category] = unique_records

    print(f"\nCategory: {category}")
    print("Original Records:", len(records))
    print("Cleaned Records:", len(unique_records))
    print("Removed Duplicates:", len(records) - len(unique_records))

print("\n========== FINAL SUMMARY ==========")

print("Before:", sum(len(x) for x in datasets.values()))
print("After:", sum(len(x) for x in datasets_dedup.values()))
print("Removed:", sum(len(x) for x in datasets.values()) -
      sum(len(x) for x in datasets_dedup.values()))
```

```python
import json

CHUNK_SIZE = 1000
CHUNK_OVERLAP = 100

chunks = []

for category, records in datasets_dedup.items():

    for record_index, record in enumerate(records):

        record_name = (
            record.get("name")
            or record.get("clean_name")
            or record.get("food_name")
            or f"{category}_{record_index}"
        )

        text = json.dumps(
            record,
            ensure_ascii=False,
            indent=2
        )

        start = 0
        chunk_index = 0

        while start < len(text):

            end = min(start + CHUNK_SIZE, len(text))

            chunk_text = text[start:end]

            content = (
                f"Category: {category}\n"
                f"Food Name: {record_name}\n\n"
                f"{chunk_text}"
            )

            chunks.append({
                "id": f"{category}_{record_index}_{chunk_index}",
                "category": category,
                "record_id": record_index,
                "record_name": record_name,
                "chunk_id": chunk_index,
                "content": content
            })

            if end >= len(text):
                break

            start = end - CHUNK_OVERLAP
            chunk_index += 1

print("Total Records:", sum(len(x) for x in datasets_dedup.values()))
print("Chunk Size:", CHUNK_SIZE)
print("Chunk Overlap:", CHUNK_OVERLAP)
print("Total Chunks:", len(chunks))

print("\nSample Chunk:")
print(chunks[0]["content"])
```

```python
import numpy as np

embedding_file = "food_embeddings (2).npy"

emb = np.load(embedding_file, mmap_mode="r")

print("Embedding Shape:", emb.shape)

total_valid = 0
total_invalid = 0

for start in range(0, len(emb), 4096):

    batch = np.asarray(emb[start:start + 4096])

    norms = np.linalg.norm(batch, axis=1)

    valid = (
        np.isfinite(batch).all(axis=1)
        & np.isclose(norms, 1.0, atol=1e-3)
    )

    total_valid += valid.sum()
    total_invalid += (~valid).sum()

print("\nValid Embeddings:", total_valid)
print("Invalid Embeddings:", total_invalid)
print("Total Embeddings:", len(emb))
```

```python
import chromadb

# Create persistent ChromaDB
client = chromadb.PersistentClient(
    path="./chroma_food_db"
)

# Create collection
collection = client.get_or_create_collection(
    name="indian_food_bilingual_v1",
    metadata={"hnsw:space": "cosine"},
    embedding_function=None
)

print("ChromaDB Initialized Successfully!")
print("Collection Name:", collection.name)
print("Existing Records:", collection.count())
```

```python
import json
import numpy as np

# Load chunks
with open("food_chunks.json", "r", encoding="utf-8") as f:
    chunks = json.load(f)

# Load validated embeddings
embeddings = np.load(
    "food_embeddings (2).npy",
    mmap_mode="r"
)

# Verify alignment
assert len(chunks) == embeddings.shape[0], "Chunk count mismatch!"
assert embeddings.shape[1] == 768, "Embedding dimension mismatch!"

BATCH_SIZE = 500
total = len(chunks)

print("Starting ChromaDB Storage...")

for i in range(0, total, BATCH_SIZE):

    batch_chunks = chunks[i:i + BATCH_SIZE]
    batch_embeddings = np.asarray(
        embeddings[i:i + BATCH_SIZE],
        dtype=np.float32
    )

    ids = [c["id"] for c in batch_chunks]

    documents = [c["content"] for c in batch_chunks]

    metadatas = [
        {
            "category": str(c["category"]),
            "record_id": int(c["record_id"]),
            "record_name": str(c["record_name"]),
            "chunk_id": int(c["chunk_id"])
        }
        for c in batch_chunks
    ]

    collection.upsert(
        ids=ids,
        embeddings=batch_embeddings.tolist(),
        documents=documents,
        metadatas=metadatas
    )

    end = min(i + BATCH_SIZE, total)

    print(f"Stored: {end}/{total}")

print("\nChromaDB Storage Completed!")
print("Total Stored Vectors:", collection.count())
```

```python
def test_retrieval(question):

    query_embedding = embedding_model.encode(
        ["query: " + question],
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    results = collection.query(
        query_embeddings=query_embedding.tolist(),
        n_results=5,
        include=["documents", "metadatas", "distances"]
    )

    print("\nQuestion:", question)
    print("=" * 60)

    for i in range(len(results["ids"][0])):

        print("\nResult:", i + 1)
        print("Food Name:", results["metadatas"][0][i]["record_name"])
        print("Category:", results["metadatas"][0][i]["category"])
        print("Distance:", results["distances"][0][i])
        print("Content:")
        print(results["documents"][0][i][:500])


# English test
test_retrieval("How to prepare chicken biryani?")

# Tamil test
test_retrieval("சிக்கன் பிரியாணி செய்வது எப்படி?")
```

```python
def retrieve_complete_recipe(question, n_results=5):

    query_embedding = embedding_model.encode(
        ["query: " + question],
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    results = collection.query(
        query_embeddings=query_embedding.tolist(),
        n_results=n_results,
        include=["metadatas", "distances"]
    )

    if not results["ids"][0]:
        print("No recipe found!")
        return

    # Get the best matching recipe
    metadata = results["metadatas"][0][0]

    category = metadata["category"]
    record_id = metadata["record_id"]
    record_name = metadata["record_name"]

    # Retrieve all chunks of that recipe
    complete_record = collection.get(
        where={
            "$and": [
                {"category": category},
                {"record_id": record_id}
            ]
        },
        include=["documents", "metadatas"]
    )

    # Sort chunks in original order
    combined = sorted(
        zip(
            complete_record["metadatas"],
            complete_record["documents"]
        ),
        key=lambda x: x[0]["chunk_id"]
    )

    print("Recipe:", record_name)
    print("Category:", category)
    print("Total Chunks:", len(combined))
    print("\nComplete Recipe Context:\n")

    for metadata, document in combined:
        print(document)
        print("-" * 50)


retrieve_complete_recipe(
    "How to prepare chicken biryani?"
)
```

```python
def get_recipe_context(question, n_results=5):

    # Convert question into embedding
    query_embedding = embedding_model.encode(
        ["query: " + question],
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    # Search ChromaDB
    results = collection.query(
        query_embeddings=query_embedding.tolist(),
        n_results=n_results,
        include=["metadatas", "distances"]
    )

    if not results["ids"][0]:
        return None

    # Get best matching record
    best = results["metadatas"][0][0]

    category = best["category"]
    record_id = best["record_id"]
    record_name = best["record_name"]

    # Retrieve all chunks of selected record
    data = collection.get(
        where={
            "$and": [
                {"category": category},
                {"record_id": record_id}
            ]
        },
        include=["documents", "metadatas"]
    )

    # Sort chunks in original order
    sorted_chunks = sorted(
        zip(data["metadatas"], data["documents"]),
        key=lambda x: x[0]["chunk_id"]
    )

    # Combine complete context
    context = "\n\n".join(
        document for metadata, document in sorted_chunks
    )

    return {
        "record_name": record_name,
        "category": category,
        "context": context,
        "total_chunks": len(sorted_chunks)
    }


# Test
recipe = get_recipe_context(
    "How to prepare chicken biryani?"
)

if recipe:
    print("Recipe:", recipe["record_name"])
    print("Category:", recipe["category"])
    print("Total Chunks:", recipe["total_chunks"])
    print("Context Length:", len(recipe["context"]))
    print("\nContext Preview:")
    print(recipe["context"][:3000])
```
